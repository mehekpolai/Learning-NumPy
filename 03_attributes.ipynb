#### 2. shape attribute:

→ shape is a predefined attribute in numpy array.

→ We should access this shape attribute by using numpy array object-

→ By using this we can check number of rows and columns in an array.

→ shape attribute returns the tuple as number of rowsw and columns.


```python
# Array with group of values
import numpy as np

df = np.array([[10, 20, 30],
               [40, 50, 60]])

print(df)
print(df.shape)
```

#### 3. ndim attribute:
→ It is a predefined attribute in numpy array.

→ We should access this ndim attribute by using numpy array object.

→ By using this we can check the dimensions of an array.

```python
# creating numpy array,check with ndim attribute
import numpy as np
data=[10,20,30,40,50]
sales=np.array(data)
print(sales.ndim)
```

```python
# if it is a nested list and then check ndim
import numpy as np
data=[[10,20],[30,40]]
sales=np.array(data)
print(sales.ndim)
```

```python
# if it is a nested list and then check ndim
import numpy as np
data=[[10,20],[30,40],[100,200]]
sales=np.array(data)
print(sales.ndim)
```

#### 4. arrayobject.T
→ T is a predefined attribute in array.

→ We should acress this T attribute by using numpy array object.

→ By using this we can transpose the array means it converts  rows as columns and columns as rows.

```python
# T attribute
import numpy as np
data=[[10,20],[30,40]]
sales=np.array(data)
print(sales)
print(sales.T)
```

```python
# T attribute
import numpy as np
data=[[10,20,20],[30,40,40]]
sales=np.array(data)
print(sales)
print(sales.T)
```
