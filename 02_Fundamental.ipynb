#### 1.Creating numpy array:

#### 2) numpy-ndim:

→ Ndim is predefined variable in numpy.
→ By using this we can check the array dimensions.


```python
# array with single value
import numpy as np
age=16
a=np.array(age)
print(age,type(age))
print(a,type(a))
```

```python
# array with multiple value
import numpy as np
ages=[16,33,31,12]
a=np.array(age)
print(ages,type(ages))
print(a,type(a))
```

```python
# array with nested list
import numpy as np
ages=[16,33,31,12,[74,10,50]]
a=np.array(age)
print(ages,type(ages))
print(a,type(a))
```

```python
# array with group of values
import numpy as np
ages=[[16,33],[31,12],[74,10,50]]
a=np.array(age)
print(ages,type(ages))
print(a,type(a))
```

#### 3) Indexing and slicing:

→ We can access numpy array values by using Indexing and slicing.
→ Numpy array having indexing nature.

→ Numpy array start with 0.

   * First element stores in Oth index.

   * Second element stores in 1st index etc.

→ By using slicing we can access piece of array from the main array.

```python
# accessing numpy by using indexing
import numpy as np
details=[10,20,30,40,50]
sales=np.array(details)
print(sales)
print(sales[0])
print(sales[1])
print(sales[6])
```

```python
# accessing numpy by using slicing
import numpy as np
details=[10,20,30,40,50]
sales=np.array(details)
print(sales)
print(sales[0:3])
print(sales[1:4])
```

```python
# Create matrix and select elements
import numpy as np

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(matrix)

print(matrix[0, 0])
print(matrix[0, 1])
print(matrix[0, 2])
print(matrix[1, 2])
print(matrix[2, 2])

```

#### Index Error:
→ If we try to access value with out of bounds of index then we will get IndexError.

```python
import numpy as np
details=[10,20,30,40,50]
sales=np.array(details)
print(sales)
print(sales[22])
```

#### 4. Creating a array with all zeros:

→ We can create array with all zeros by using numpy.zeros() function.



```python
# create a numpy with group of values
import numpy as np
sales=np.zeros(5)
print(sales)
print(type(sales))
```
